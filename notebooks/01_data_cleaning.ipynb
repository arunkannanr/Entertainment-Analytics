## 4. Removing Duplicate Records from Spotify

```python
spotify_cleaned = spotify.drop_duplicates()
```

```python
spotify_cleaned.duplicated().sum()
```

```python
spotify.shape
```

```python
spotify_cleaned.shape
```

```python
netflix_missing = netflix.isnull().mean() * 100
netflix_missing.sort_values(ascending=False)
```

```python
netflix["director"] = netflix["director"].fillna("Unknown")
netflix["cast"] = netflix["cast"].fillna("Unknown")
netflix["country"] = netflix["country"].fillna("Unknown")
netflix["rating"] = netflix["rating"].fillna("Unknown")
netflix["duration"] = netflix["duration"].fillna("Unknown")
```

```python
netflix["date_added"].dtype
```

```python
netflix["date_added"] = pd.to_datetime(netflix["date_added"], errors="coerce")
```

```python
netflix.isnull().sum()
```

```python
spotify_cleaned["artists"] = spotify_cleaned["artists"].fillna("Unknown")
spotify_cleaned["album_name"] = spotify_cleaned["album_name"].fillna("Unknown")
spotify_cleaned["track_name"] = spotify_cleaned["track_name"].fillna("Unknown")
```
